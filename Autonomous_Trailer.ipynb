# 1.Identify the Task
The task is to develop an autonomous trailer that can navigate from one point to another e.g from Nairobi to Kisumu while making decisions based on road conditions, traffic and safety requirements.

Goal: Reach Kisumu using the shortest route while ensuring safe travel.

# 2. Assemble Relevant Knowledge
The system needs to know:
       The available routes between locations
       The distance between each location
       The road conditions and traffic situation
       The safety requirements for travelling
       The possible actions the trailer can take
No external APIs or web scraping tools are used.

# 3.Define the Vocabulary
Predicates (represented in Python as keys and rules): Route(from, to), Distance(from, to, km), Safe(route)

Constants: Nairobi, Naivasha, Nakuru, Kericho, Kisumu

# 4.Encode General Knowledge about the Domain

```python
rules = {
    "Nairobi": ["Naivasha"],
    "Naivasha": ["Nakuru"],
    "Nakuru": ["Kericho", "Kisumu"],
    "Kericho": ["Kisumu"]
}

distances = {
    ("Nairobi", "Naivasha"): 92.8,
    ("Naivasha", "Nakuru"): 71.1,
    ("Nakuru", "Kericho"): 107,
    ("Nakuru", "Kisumu"): 191,
    ("Kericho", "Kisumu"): 83.9
}
```

# 5: Add New Facts to the Knowledge Base 


```python
def show_kb():
    print("\n--- Current Knowledge Base ---")
    for location, routes in rules.items():
        print(f"{location:10} ➜ {routes}")
    print("-------------------------------")

def add_new_fact(location, route, distance):
    rules.setdefault(location, []).append(route)
    distances[(location, route)] = distance
    print(f"Added fact: '{location}' ➜ '{route}' ({distance} km)")

def add_fact():
    new_location = input("Enter starting location: ").title()
    new_route = input("Enter destination location: ").title()
    new_distance = float(input("Enter distance in km: "))

    rules.setdefault(new_location, []).append(new_route)
    distances[(new_location, new_route)] = new_distance

    print(f"New fact added: '{new_location}' ➜ '{new_route}' ({new_distance} km)")

add_fact()
show_kb()
```

# 6: Encode Specific Problem Instance (Facts) & Pose Queries and Get Answers

```python
# Step 6: Current journey details
current_location = input("Enter current location: ").title()
destination = input("Enter destination: ").title()

# Step 7: Query the knowledge base
def get_routes(location):
    available_routes = rules.get(location, [])
    return [
        (route, distances[(location, route)])
        for route in available_routes
        if (location, route) in distances
    ]

print("\n=== Autonomous Trailer Navigation ===")
print("Current Location:", current_location)
print("Destination:", destination)
print("Available Routes:", get_routes(current_location))
```

# 8: Querying the Knowledge Base

```python
# Ask questions to retrieve routes from known facts
print("Routes from Nairobi:", get_routes("Nairobi"))
print("Routes from Nakuru:", get_routes("Nakuru"))
print("Routes from Kericho:", get_routes("Kericho"))
print("Routes from Kisumu:", get_routes("Kisumu"))
```

# 9: Debugging the Knowledge Base

```python
# Test known and unknown locations
test_cases = ["Nairobi", "Naivasha", "Nakuru", "Kericho", "Kisumu", "Mombasa", "Eldoret"]

print("\n--- Debugging the Knowledge Base ---")

for location in test_cases:
    print(f"Location: {location:10} -> Routes: {get_routes(location)}")
```

# 10: Final Reflection

The knowledge base helps the autonomous trailer identify available routes and their distances. One challenge is that the system relies on predefined route information and does not account for real-time traffic or road conditions. Future improvements could include integrating GPS, traffic updates and road condition sensors to improve navigation and safety.
